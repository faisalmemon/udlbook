```python
import numpy as np
import matplotlib.pyplot as plt
import networkx as nx
```

```python
# Adjacency matrix for graph A1
adj_matrix_a1 = np.array([
    [0, 1, 1, 1, 0, 1, 1],  # Node 1
    [1, 0, 1, 1, 1, 0, 0],  # Node 2
    [1, 1, 0, 0, 0, 0, 0],  # Node 3
    [1, 1, 0, 0, 1, 0, 0],  # Node 4
    [0, 1, 0, 1, 0, 0, 0],  # Node 5
    [1, 0, 0, 0, 0, 0, 1],  # Node 6
    [1, 0, 0, 0, 0, 1, 0],  # Node 7
])

# Adjacency matrix for graph A2
adj_matrix_a2 = np.array([
    [0, 0, 0, 1, 0, 0, 1, 1],  # Node 1
    [0, 0, 1, 0, 1, 1, 0, 0],  # Node 2
    [0, 1, 0, 0, 0, 0, 0, 1],  # Node 3
    [1, 0, 0, 0, 0, 0, 0, 1],  # Node 4
    [0, 1, 0, 0, 0, 1, 1, 0],  # Node 5
    [0, 1, 0, 0, 1, 0, 1, 0],  # Node 6
    [1, 0, 0, 0, 1, 1, 0, 0],  # Node 7
    [1, 0, 1, 1, 0, 0, 0, 0],  # Node 8
])

# Compute powers of adjacency matrices
a1_cubed = np.linalg.matrix_power(adj_matrix_a1, 3)
a1_seventh = np.linalg.matrix_power(adj_matrix_a1, 7)
a2_cubed = np.linalg.matrix_power(adj_matrix_a2, 3)
a2_seventh = np.linalg.matrix_power(adj_matrix_a2, 7)

# One-hot vectors for node 1
node1_a1_onehot = np.array([1, 0, 0, 0, 0, 0, 0]).reshape(7, 1)
node1_a2_onehot = np.array([1, 0, 0, 0, 0, 0, 0, 0]).reshape(8, 1)

# Compute neighbors reachable in 3 and 7 hops
a1_hop3_node1 = np.matmul(a1_cubed, node1_a1_onehot)
a1_hop7_node1 = np.matmul(a1_seventh, node1_a1_onehot)
a2_hop3_node1 = np.matmul(a2_cubed, node1_a2_onehot)
a2_hop7_node1 = np.matmul(a2_seventh, node1_a2_onehot)

# Print results
print("A1: 3-hop neighbors from Node 1 to Node 2 (Expected: 9):", a1_hop3_node1[1, 0])
print("A1: 7-hop neighbors from Node 1 to Node 2 (Expected: 879):", a1_hop7_node1[1, 0])
print("A2: 3-hop neighbors from Node 1 to Node 2 (Expected: 3):", a2_hop3_node1[1, 0])
print("A2: 7-hop neighbors from Node 1 to Node 2 (Expected: 203):", a2_hop7_node1[1, 0])
```
