```python
import numpy as np
import matplotlib.pyplot as plt
import networkx as nx
```

```python
# Adjacency matrix for graph A1
a1 = np.array([
    #1  2  3  4  5  6  7
    [0, 1, 1, 1, 0, 1, 1],  # Node 1
    [1, 0, 1, 1, 1, 0, 0],  # Node 2
    [1, 1, 0, 0, 0, 0, 0],  # Node 3
    [1, 1, 0, 0, 1, 0, 0],  # Node 4
    [0, 1, 0, 1, 0, 0, 0],  # Node 5
    [1, 0, 0, 0, 0, 0, 1],  # Node 6
    [1, 0, 0, 0, 0, 1, 0],  # Node 7
])

# Adjacency matrix for graph A2
a2 = np.array([
    #1  2  3  4  5  6  7  8
    [0, 0, 0, 1, 0, 0, 1, 1],  # Node 1
    [0, 0, 1, 0, 1, 1, 0, 0],  # Node 2
    [0, 1, 0, 0, 0, 0, 0, 1],  # Node 3
    [1, 0, 0, 0, 0, 0, 0, 1],  # Node 4
    [0, 1, 0, 0, 0, 1, 1, 0],  # Node 5
    [0, 1, 0, 0, 1, 0, 1, 0],  # Node 6
    [1, 0, 0, 0, 1, 1, 0, 0],  # Node 7
    [1, 0, 1, 1, 0, 0, 0, 0],  # Node 8
])

a1_3 = np.linalg.matrix_power(a1, 3)
a1_7 = np.linalg.matrix_power(a1, 7)
a2_3 = np.linalg.matrix_power(a2, 3)
a2_7 = np.linalg.matrix_power(a2, 7)

node_1_a1_onehot = np.array([1, 0, 0, 0, 0, 0, 0]).reshape(7, 1)
node_1_a2_onehot = np.array([1, 0, 0, 0, 0, 0, 0, 0]).reshape(8, 1)
a1_hop3_from_node_1 = np.matmul(a1_3, node_1_a1_onehot)
print("Three hop neighbors from A1 node 1 to 2:==Want 9==")
print(a1_hop3_from_node_1[1])

a1_hop7_from_node_1 = np.matmul(a1_7, node_1_a1_onehot)
print("7 hops neighbors from A1 node 1 to 2:==Want 879==")
print(a1_hop7_from_node_1[1])

a2_hop3_from_node_1 = np.matmul(a2_3, node_1_a2_onehot)
print("3 hops neighbors from A2 node 1 to 2:==Want 3==")
print(a2_hop3_from_node_1[1])
a2_hop7_from_node_1 = np.matmul(a2_7, node_1_a2_onehot)
print("7 hops neighbors from node 1 in A2:==Want 203==")
print(a2_hop7_from_node_1[1])
```
