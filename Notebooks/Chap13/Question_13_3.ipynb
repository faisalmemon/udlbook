````python
import numpy as np
import matplotlib.pyplot as plt
import networkx as nx
````

````python
# Adjacency matrix for graph A1
a1 = np.array([
    [0, 1, 1, 0, 0, 1, 1],  # Node 1
    [1, 0, 1, 1, 1, 0, 0],  # Node 2
    [1, 1, 0, 0, 0, 0, 0],  # Node 3
    [0, 1, 0, 0, 1, 0, 0],  # Node 4
    [0, 1, 0, 1, 0, 0, 0],  # Node 5
    [1, 0, 0, 0, 0, 0, 1],  # Node 6
    [1, 0, 0, 0, 0, 1, 0],  # Node 7
])


# update this
a2 = np.array([
    [0, 0, 0, 1, 0, 0, 1, 1],  # Node 1
    [0, 0, 1, 0, 1, 1, 0, 0],  # Node 2
    [0, 1, 0, 0, 0, 0, 0, 1],  # Node 3
    [1, 0, 0, 0, 0, 0, 0, 1],  # Node 4
    [0, 1, 0, 0, 0, 1, 1, 0],  # Node 5
    [0, 1, 0, 0, 1, 0, 1, 0],  # Node 6
    [1, 0, 0, 0, 1, 1, 0, 0],  # Node 7
    [1, 0, 1, 1, 0, 0, 0, 0],  # Node 8
])

#print("Original matrix A1:")
#print(a1)

a1_2 = np.multiply(a1, a1)
a1_3 = np.linalg.matrix_power(a1, 3)

#print("Matrix A1^3:")
#print(a1_3)

a1_7 = np.linalg.matrix_power(a1, 7)
#print("Matrix A1^7:")
#print(a1_7)

node_1_onehot = np.array([1, 0, 0, 0, 0, 0, 0]).reshape(7, 1)
node_2_onehot = np.array([0, 1, 0, 0, 0, 0, 0, 0]).reshape(8, 1)
three_hop_neighbors_from_node_1 = np.matmul(a1_3, node_1_onehot)
print("Three hop neighbors from node 1:==9==")
print(three_hop_neighbors_from_node_1)

hop_7_times_neighbors_from_node_1 = np.matmul(a1_7, node_1_onehot)
print("7 hops neighbors from node 1:==879==")
print(hop_7_times_neighbors_from_node_1)

a2_3 = np.linalg.matrix_power(a2, 3)
a2_7 = np.linalg.matrix_power(a2, 7)

hop_3_neighbours_from_node_1 = np.matmul(a2_3, node_2_onehot)
print("3 hops neighbors from node 1 in A2:==3==")
print(hop_3_neighbours_from_node_1)
hop_7_neighbours_from_node_1 = np.matmul(a2_7, node_2_onehot)
print("7 hops neighbors from node 1 in A2:==203==")
print(hop_7_neighbours_from_node_1)
````
